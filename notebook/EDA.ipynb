```python
## all the necessary libaries imported 

import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
import warnings
warnings.filterwarnings('ignore')
```

```python
## importing csv file
df = pd.read_csv('data/apple_global_sales_dataset.csv')
## looking at top 5 records
df.head(5)
df.shape ### we have 27 featuers and 11500 records
```

### 3. Data Checks to perform

- Check Missing values
- Check Duplicates
- Check data type
- Check the number of unique values of each column
- Check statistics of data set
- Check various categories present in the different categorical column

```python
## Check Misssing Values
##these are the features with nan value
feature_with_na = [feature for feature in df.columns if df[feature].isnull().sum()]
for feature in feature_with_na:
    print(feature,np.round(df[feature].isnull().mean()*100,5),'% missing value')
```

```python
## storage of products like watch and airposd missing so writing unapplicable in those
df['storage'] = df['storage'].fillna('Not Applicable')
##imputing the na values with "unknown"
df['previous_device_os'] = df['previous_device_os'].fillna('Unknown')
## since there are lot of values missing in customer rating we are dropping this feature
df.drop(columns=['customer_rating'],inplace=True)
```

```python
## we have no duplicate values
df.duplicated().sum()
```

```python
numeric_features = [feature for feature in df.columns if df[feature].dtype != 'O']
categorical_features = [feature for feature in df.columns if df[feature].dtype == 'O']

print('We have {} numerical features : {}'.format(len(numeric_features), numeric_features))
print('\nWe have {} categorical features : {}'.format(len(categorical_features), categorical_features))


## we had a total of 27 featuers, we dropped customer_rating feature because of 29% missing values now we have 8 numerical feature and 18 categorical features
```

```python
df.columns
df[categorical_features]
# sale_id needs to be dropped as well
# need segregate sale_date to day month and year and then drop sale_date because quater and month are already being covered
# also currency can be droppen since we have a feature of country
# city can also be dropped since we are having featuers like country, region 
# fx_rate_to_usd and revenue_local_currency these are just currency conversion artifacts.
# discounted_price_usd carries zero new information that unit_price_usd and discount_pct together don't already tell the model.

```

```python
df['sale_date'] = pd.to_datetime(df['sale_date']0)
df['day'] = df['sale_date'].dt.day
df['day_of_week'] = df['sale_date'].dt.day_of_week
```

```python
df.drop(columns=[
    'sale_id',           
    'sale_date',         
    'discounted_price_usd',      
    'revenue_local_currency',   
    'fx_rate_to_usd',           
    'currency'           
], inplace=True)
```

```python
df.drop(columns=['city'], inplace= True)
```
